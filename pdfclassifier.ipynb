```python
import pandas as pd
import numpy as np
import os
import re
from PyPDF2 import PdfReader
from sklearn.feature_extraction.text import TfidfVectorizer
import string
```

PreProcessing of text 

```python
def extract_text_from_pdf(pdf_path):
    text = ""
    reader = PdfReader(pdf_path)
    for page in reader.pages:
        text += page.extract_text()
    return text

def extract_text_from_txt(txt_path):
    with open(txt_path, 'r', encoding='utf-8', errors='ignore') as file:
        text = file.read()
    return text

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'\s+', ' ', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = re.sub(r'\d+', '', text)
    text = text.strip()
    return text

data = []
labels = []

dataset_path = r'c:\Users\hp\Desktop\coding\SUPERVISED\DATASET'

for class_name in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_name)
    if os.path.isdir(class_path):
        for filename in os.listdir(class_path):
            file_path = os.path.join(class_path, filename)
            if filename.endswith('.pdf'):
                text = extract_text_from_pdf(file_path)
            elif filename.endswith('.txt'):
                text = extract_text_from_txt(file_path)
            else:
                continue
            
            processed_text = preprocess_text(text)
            data.append(processed_text)
            labels.append(class_name)

df = pd.DataFrame({'text': data, 'label': labels})
```

```python
def extract_with_pypdf(pdf_path):
    text = ""
    try:
        reader = PdfReader(pdf_path)
        for page in reader.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
    except Exception as e:
        return ""
    
    if len(text.strip()) < 50:
        return ""
    
    return text

def extract_from_txt(txt_path):
    try:
        with open(txt_path, 'r', encoding='utf-8', errors='ignore') as file:
            text = file.read()
        return text
    except Exception as e:
        return ""

def process_all_files(base_directory):
    documents, labels, filenames = [], [], [] 
    
    if not os.path.isdir(base_directory):
        return documents, labels, filenames

    all_classes = [d for d in os.listdir(base_directory) if os.path.isdir(os.path.join(base_directory, d))]
    
    for class_name in all_classes:
        class_path = os.path.join(base_directory, class_name)
        
        all_files = [f for f in os.listdir(class_path) if f.lower().endswith((".pdf", ".txt"))]
        
        for filename in all_files:
            path = os.path.join(class_path, filename)
            
            if filename.lower().endswith('.pdf'):
                text = extract_with_pypdf(path)
            elif filename.lower().endswith('.txt'):
                text = extract_from_txt(path)
            else:
                continue
            
            if text and len(text.strip()) >= 50:
                documents.append(text)
                labels.append(class_name)
                filenames.append(filename)
    
    return documents, labels, filenames
```

```python
TRAINING_DOCS_DIR = r"C:\Users\hp\Desktop\coding\SUPERVISED\DATASET"
documents, labels, all_filenames = process_all_files(TRAINING_DOCS_DIR)
```

```python
from transformers import BertTokenizer

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

tokenized_data = []
for text in data:
    tokens = tokenizer(
        text,
        padding='max_length',
        truncation=True,
        max_length=512,
        return_tensors='pt'
    )
    tokenized_data.append(tokens)
    
```

```python
import torch
from transformers import BertModel
```

```python
model = BertModel.from_pretrained('bert-base-uncased')
model.eval()

embeddings = []
with torch.no_grad():
    for tokens in tokenized_data:
        outputs = model(
            input_ids=tokens['input_ids'],
            attention_mask=tokens['attention_mask']
        )
        cls_embedding = outputs.last_hidden_state[:, 0, :]
        embeddings.append(cls_embedding.squeeze().numpy())

embeddings = np.array(embeddings)
print(embeddings)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
```

```python
# Use the documents and labels from the same processing pipeline
tokenized_documents = []
for text in documents:
    tokens = tokenizer(
        text,
        padding='max_length',
        truncation=True,
        max_length=512,
        return_tensors='pt'
    )
    tokenized_documents.append(tokens)

# Generate embeddings for the documents
document_embeddings = []
with torch.no_grad():
    for tokens in tokenized_documents:
        outputs = model(
            input_ids=tokens['input_ids'],
            attention_mask=tokens['attention_mask']
        )
        cls_embedding = outputs.last_hidden_state[:, 0, :]
        document_embeddings.append(cls_embedding.squeeze().numpy())

document_embeddings = np.array(document_embeddings)

label_encoder = LabelEncoder()
encoded_labels = label_encoder.fit_transform(labels)

X_train, X_test, y_train, y_test = train_test_split(
    document_embeddings, encoded_labels, test_size=0.2, random_state=42, stratify=encoded_labels
)

X_train_tensor = torch.FloatTensor(X_train)
y_train_tensor = torch.LongTensor(y_train)
X_test_tensor = torch.FloatTensor(X_test)
y_test_tensor = torch.LongTensor(y_test)

train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
test_dataset = TensorDataset(X_test_tensor, y_test_tensor)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)

class Classifier(nn.Module):
    def __init__(self, input_dim, num_classes):
        super(Classifier, self).__init__()
        self.fc1 = nn.Linear(input_dim, 256)
        self.dropout1 = nn.Dropout(0.3)
        self.fc2 = nn.Linear(256, 128)
        self.dropout2 = nn.Dropout(0.3)
        self.fc3 = nn.Linear(128, num_classes)
        self.relu = nn.ReLU()
        
    def forward(self, x):
        x = self.relu(self.fc1(x))
        x = self.dropout1(x)
        x = self.relu(self.fc2(x))
        x = self.dropout2(x)
        x = self.fc3(x)
        return x

num_classes = len(label_encoder.classes_)
classifier = Classifier(embeddings.shape[1], num_classes)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(classifier.parameters(), lr=0.001)

for epoch in range(15):
    classifier.train()
    train_loss = 0.0
    
    for inputs, labels_batch in train_loader:
        optimizer.zero_grad()
        outputs = classifier(inputs)
        loss = criterion(outputs, labels_batch)
        loss.backward()
        optimizer.step()
        train_loss += loss.item()
    
    classifier.eval()
    correct = 0
    total = 0
    
    with torch.no_grad():
        for inputs, labels_batch in test_loader:
            outputs = classifier(inputs)
            _, predicted = torch.max(outputs, 1)
            total += labels_batch.size(0)
            correct += (predicted == labels_batch).sum().item()
    
    accuracy = 100 * correct / total
    print(f'Epoch [{epoch+1}/15], Loss: {train_loss/len(train_loader):.4f}, Accuracy: {accuracy:.2f}%')
```
